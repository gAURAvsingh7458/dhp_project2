```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Load dataset
dataset = pd.read_csv('Clustering_Customers.csv')

# Select Annual Income and Spending Score
X = dataset.iloc[:, [3, 4]].values

# Calculate WCSS
wcss = []

for i in range(1, 50):
    kmeans = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=1,
        n_init=10
    )

    kmeans.fit(X)
    wcss.append(kmeans.inertia_)


plt.plot(range(1, 50), wcss, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.xticks(range(1, 50))
plt.show()
```

```python
kmeans=KMeans(n_clusters=5,init="k-means++",random_state=1)
y_means=kmeans.fit_predict(X)
```

```python
plt.scatter(X[y_means == 0, 0], X[y_means == 0, 1], s=100, c='red', label='c1')
plt.scatter(X[y_means == 1, 0], X[y_means == 1, 1], s=100, c='cyan', label='c2')
plt.scatter(X[y_means == 2, 0], X[y_means == 2, 1], s=100, c='magenta', label='c3')
plt.scatter(X[y_means == 3, 0], X[y_means == 3, 1], s=100, c='black', label='c4')
plt.scatter(X[y_means == 4, 0], X[y_means == 4, 1], s=100, c='purple', label='c5')

plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
 
```
